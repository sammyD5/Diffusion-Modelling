```python
import sympy as sp
# Sympy 
x = sp.symbols('x')
equation = sp.Eq(x**2 - 5*x +6, 0)
solutions = sp.solve(equation, x)

print("solutions:", solutions)
sp.latex(equation)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
expr = x**2 + 2*x + 1
simplified_expr = sp.simplify(expr)
print(simplified_expr)  # Output: (x + 1)**2
sp.latex(expr)  # Displays the expression in a pretty format
```
